```python
def get_phase(overs):
    if overs <= 5:
        return 'Powerplay'
    elif overs <= 15:
        return 'Middle Overs'
    else:
        return 'Death Overs'
df['phase'] = df['overs'].apply(get_phase)

all_phases_sample = df.groupby('phase').head(1).sort_values('overs')
display(all_phases_sample[['overs', 'phase']])
```

```python
team_runs = df.groupby('batting_team')['total_run'].sum().sort_values(ascending=False).reset_index()

plt.figure(figsize=(12, 8))
sns.barplot(data=team_runs, x='total_run', y='batting_team', palette='coolwarm')
plt.title('Total Runs Scored by Each Team (2008-2024)')
plt.xlabel('Total Runs')
plt.ylabel('Team')
plt.show()
```

```python
batter_stats = df.groupby('batter').agg({'batsman_run': 'sum', 'ball_number': 'count'})
batter_stats['strike_rate'] = (batter_stats['batsman_run'] / batter_stats['ball_number']) * 100

top_sr = batter_stats[batter_stats['ball_number'] > 500].sort_values('strike_rate', ascending=False).head(10).reset_index()

plt.figure(figsize=(12, 6))
sns.barplot(data=top_sr, x='strike_rate', y='batter', palette='magma')
plt.title('Top 10 Batters by Strike Rate (Min. 500 balls)')
plt.xlabel('Strike Rate')
plt.ylabel('Batter')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.countplot(data=df[df['iswicket_delivery'] == 1], x='dismisal_kind', order=df['dismisal_kind'].value_counts().index, palette='viridis')
plt.title('Distribution of Dismissal Types')
plt.xticks(rotation=45)
plt.xlabel('Type of Dismissal')
plt.ylabel('Frequency')
plt.show()
```

### **Final Insights Summary (2008-2024)**

After exploring the ball-by-ball IPL dataset, several key trends emerge:

*   **Consistency**: Virat Kohli and Shikhar Dhawan lead the league in total runs and fours, respectively, proving their value as long-term anchors.
*   **High Impact**: Andre Russell and Heinrich Klaasen represent the modern T20 era with the highest strike rates among established batters.
*   **Bowling Dominance**: Yuzvendra Chahal and Dwayne Bravo lead the wicket-taking charts, with 'Caught' being the most frequent dismissal type.
*   **Game Dynamics**: While dot balls are the most common outcome of a delivery, the ability to hit sixes (led by Chris Gayle) remains the biggest differentiator in winning matches.
