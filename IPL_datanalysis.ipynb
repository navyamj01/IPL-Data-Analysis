```python
total_runs_per_batter=df.groupby('batter')['batsman_run'].sum().sort_values(ascending=False).reset_index()
display(total_runs_per_batter.head(10))
```

```python
top_10_batters=total_runs_per_batter.head(10)

plt.figure(figsize=(12,6))
sns.barplot(data=top_10_batters, x='batsman_run',y='batter',palette='viridis')
plt.title('Top 10 batters from 2008-2024')
plt.xlabel('batsman_run')
plt.ylabel('batter')
plt.show()
```

```python
top_bowlers=df.groupby('bowler')['iswicket_delivery'].sum().sort_values(ascending=False).reset_index()
top_bowlers.head(10)
```

```python
top_10_bowlers=top_bowlers.head(10)

plt.figure(figsize=(12,6))
sns.barplot(data=top_10_bowlers, x='iswicket_delivery',y='bowler',palette='viridis')
plt.title('Top 10 bowlers from 2008-2024')
plt.xlabel('iswicket_delivery')
plt.ylabel('bowler')
plt.show()
```

```python
sixes=df[df['batsman_run']==6]
top_six_hitters=sixes['batter'].value_counts().head(10)
top_six_hitters
```

```python
top_sixers = top_six_hitters.head(10).reset_index()

plt.figure(figsize=(12,6))
sns.barplot(data=top_sixers, x='count', y='batter', palette='viridis')
plt.title('Top 10 sixers from 2008-2024')
plt.xlabel('Number of Sixes')
plt.ylabel('Batter')
plt.show()
```

```python
fours=df[df['batsman_run']==4]
top_four_hitters=fours['batter'].value_counts().head(10)
top_four_hitters
```

```python
top_fours = top_four_hitters.head(10).reset_index()

plt.figure(figsize=(12,6))
sns.barplot(data=top_fours, x='count', y='batter', palette='viridis')
plt.title('Top 10 fourers from 2008-2024')
plt.xlabel('Number of Fours')
plt.ylabel('Batter')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='batsman_run', palette='magma')
plt.title('Frequency of Runs Scored per Ball (2008-2024)')
plt.xlabel('Runs Scored')
plt.ylabel('Number of Balls')
plt.show()
```

```python
team_runs=df.groupby('batting_team')['total_run'].sum()
team_runs
```

```python
df['is_dot'] = (df['total_run'] == 0).astype(int)

dot_balls = df.groupby('bowler')['is_dot'].sum().sort_values(ascending=False).head(10)
dot_balls
```

```python
runs = df.groupby("batter")["batsman_run"].sum()
balls = df.groupby("batter")["ball_number"].count()

strike_rate = (runs / balls) * 100
strike_rate
```
